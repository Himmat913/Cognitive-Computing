### Q1

```python
# Q1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Readings:", temperature)
```

```python
# Q1 a)
corrected = temperature + 2
print("After adding 2:", corrected)
```

```python
# Q1 b)
fahrenheit = (9 / 5) * temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)
```

```python
# Q1 c)
high = temperature[temperature > 32]
print("Readings greater than 32:", high)
```

```python
# Q1 d)
count = np.sum(temperature > 32)
print("Number of readings above 32:", count)
```

**Q1 e)** With vectorization, an operation like `temperature + 2` is applied to the whole array at once. NumPy does the looping internally in pre-compiled C code, so we do not pay Python's cost of type checking and interpreting every element one by one. The code is also shorter and has fewer chances for bugs.

Boolean indexing works the same way. `temperature > 32` gives an array of True/False values in a single step, and using it as an index picks out only the matching readings. Without it we would need a `for` loop with an `if` and `append`. For millions of readings the NumPy version is much faster and easier to read.

### Q2

```python
# Q2
steps = np.array([[5000, 6200, 7100],
                  [8000, 7500, 9000],
                  [4500, 5100, 4800],
                  [9000, 8500, 9500]])
print(steps)
```

```python
# Q2 a)
print("Total steps:", np.sum(steps))
```

```python
# Q2 b)
print("Mean steps:", np.mean(steps))
```

```python
# Q2 c)
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))
```

```python
# Q2 d)
print("Total steps for each day:", np.sum(steps, axis=0))
```

```python
# Q2 e)
print("Total steps for each user:", np.sum(steps, axis=1))
```

```python
# Q2 f)
pos = np.argmax(steps)
print("Flat index of maximum:", pos)

user = pos // steps.shape[1]
day = pos % steps.shape[1]
print("User", user + 1, "on day", day + 1, "has the maximum steps:", steps[user, day])
```

`argmax()` on a 2-D array gives the position in the flattened array, so dividing by the number of columns gives the row (user) and the remainder gives the column (day).

### Q3

```python
# Q3 a)
original = np.array([1, 2, 3, 4, 5, 6])
print(original)
```

```python
# Q3 b)
subset = original[1:4]
print("subset:", subset)
```

```python
# Q3 c)
subset[0] = 999
print("original:", original)
print("subset:", subset)
```

The change in `subset` also shows up in `original`, because a basic slice is only a view of the same memory and not a separate copy.

```python
# Q3 d)
copied = original[1:4].copy()
copied[0] = 500
print("original:", original)
print("copied:", copied)
```

This time `original` is unchanged, because `.copy()` makes a new array with its own memory.

```python
# Q3 e)
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)
```

```python
# Q3 f)
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])
```

```python
# Q3 g)
flat = matrix.flatten()
rav = matrix.ravel()
print("flatten():", flat)
print("ravel():", rav)
```

```python
# Q3 h)
rav[0] = 100
print("ravel array:", rav)
print("matrix after changing ravel array:")
print(matrix)
```

The original matrix is affected, because `ravel()` returns a view of the original array.

```python
# Q3 i)
flat[1] = 200
print("flatten array:", flat)
print("matrix after changing flatten array:")
print(matrix)
```

The original matrix is not affected, because `flatten()` always returns a copy.

```python
# Q3 j)
print("Shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:", matrix.size)
print("dtype:", matrix.dtype)
```

### Q4

```python
# Q4
X = np.array([[6, 70, 3],
              [5, 50, 6],
              [8, 80, 2],
              [4, 30, 8]])
y = np.array([40, 65, 30, 85])
print(X)
print(y)
```

```python
# Q4 a)
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
# Q4 b)
print("X.T:")
print(X.T)
print("Shape of X.T:", X.T.shape)
```

**Q4 b)** The transpose swaps rows and columns. In X each row is one user and each column is one feature. In `X.T` each row is one feature (sleep, activity, stress) and each column is one user, so the shape changes from (4, 3) to (3, 4).

```python
# Q4 c)
XtX = X.T @ X
print(XtX)
print("Shape:", XtX.shape)
```

```python
# Q4 d)
print("Determinant:", np.linalg.det(XtX))
XtX_inv = np.linalg.inv(XtX)
print(XtX_inv)
```

The inverse exists only for a square matrix with a non-zero determinant. `X.T @ X` is 3 × 3 and its determinant is not zero, so `np.linalg.inv()` can be used here. X itself is 4 × 3, so its inverse cannot be taken.

```python
# Q4 e)
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Coefficients:", beta)
```

**Q4 f)** There are three coefficients, one for each feature, in the order sleep hours, activity level and stress level. Each coefficient shows how much the predicted assistance score changes when that feature goes up by 1 and the other two stay the same. A positive coefficient means the score rises with that feature, and a negative one means it falls. A larger absolute value means the feature has a bigger effect on the score. Since no column of ones was added to X, this model has no intercept term.

Here the coefficients come out as about -0.16 for sleep, 0.14 for activity and 10.06 for stress. So stress level has by far the biggest effect: each extra stress point raises the predicted assistance score by about 10. Activity has a small positive effect and sleep a very small negative one. With only 4 users this is just a small practice example, so these values should not be taken as a real finding.

```python
# Q4 g)
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("Predicted assistance score:", prediction)
```
