## Q6 - Employee dataset

```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv("employees.csv", index=False)
print(emp_df)
```

```python
#a) shape of the DataFrame
print(emp_df.shape)
```

```python
#b) summary of the DataFrame
emp_df.info()
```

```python
#c) descriptive statistics
print(emp_df.describe())
```

```python
#d) first 5 rows and last 3 rows
print(emp_df.head())
print(emp_df.tail(3))
```

```python
#e) statistics
print("Average salary:", emp_df["Salary"].mean())
print("Total bonus paid:", emp_df["Bonus"].sum())
print("Youngest employee's age:", emp_df["Age"].min())
print("Highest performance rating:", emp_df["Rating"].max())
```

```python
#f) sort by Salary in descending order
print(emp_df.sort_values("Salary", ascending=False))
```

```python
#g) categorize employees by performance rating
def get_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance_Category"] = emp_df["Rating"].apply(get_category)
print(emp_df)
```

```python
#h) check for missing values
print(emp_df.isnull().sum())
```

```python
#i) rename Employee_ID to ID
emp_df = emp_df.rename(columns={"Employee_ID": "ID"})
print(emp_df.columns.tolist())
```

```python
#j.i) employees with more than 5 years of experience
print(emp_df[emp_df["Years_of_Experience"] > 5])
```

```python
#j.ii) employees in the IT department
print(emp_df[emp_df["Department"] == "IT"])
```

```python
#k) add Tax column - 10% deduction from Salary
emp_df["Tax"] = emp_df["Salary"] * 0.10
print(emp_df)
```

```python
#l) save the modified DataFrame to a new CSV file
emp_df.to_csv("employees_modified.csv", index=False)
print("Saved as employees_modified.csv")
```
