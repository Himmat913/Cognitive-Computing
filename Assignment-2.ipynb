## Q1 - List operations

```python
#Q1(i)
roll_no = "1024170271"
L = [int(d)*10 for d in roll_no]
print(L)
```

```python
#Q1(ii)
L.append(100)   # append adds 100 at the end of the list
print(L)
L.insert(2, 200)   # insert adds 200 at index 2, pushing other elements right
print(L)
```

```python
#Q1(iii)
L.remove(200)   # remove() deletes the first matching value (200)
print(L)
L.pop()   # pop() removes the last element (100) from the list
print(L)
```

```python
#Q1(iv)
L.sort()
print(L)
L.sort(reverse=True)
print(L)
```

```python
#Q1(v)
print(L[:3])
print(L[-3:])
```

```python
#Q1(vi)
avg = sum(L)/len(L)
print("Average =", avg)
newL = [x for x in L if x > avg]
print(newL)
```

## Q2 - Tuple operations

```python
#Q2(i)
L = [int(d)*10 for d in roll_no]
scores = tuple(L[:8])
print(scores)
print("Highest:", max(scores), "at index:", scores.index(max(scores)))
print("Lowest:", min(scores), "appears:", scores.count(min(scores)), "times")
```

```python
#Q2(ii)
# tuples are immutable, so they don't have a reverse() method like lists do
rev_scores = list(scores[::-1])
print(rev_scores)
```

```python
#Q2(iii)
x = int(input("Enter the score to find: "))
if x in scores:
    print("Found", x, "at index:", scores.index(x))
else:
    print(x, "not present in scores")
```

```python
#Q2(iv)
scores[0] = 100
```

##### Explanation: tuple is immutable so its elements can't be changed after creation, unlike a list

```python
#Q2(v)
first_score, second_score, *remaining_scores = scores
print(first_score)
print(second_score)
print(remaining_scores)
```

## Q3 - Random numbers

```python
#Q3(i)
import random
random.seed(1024170271)   # my roll number, keeps the list reproducible and unique to me
randL = [random.randint(100, 900) for _ in range(100)]
print(randL)
```

```python
#Q3(ii)
oddL = [n for n in randL if n % 2 != 0]
print(oddL)
print("Count of odd numbers:", len(oddL))
```

```python
#Q3(iii)
evenL = [n for n in randL if n % 2 == 0]
print(evenL)
print("Count of even numbers:", len(evenL))
```

```python
#Q3(iv)
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num//2 + 1):
        if num % i == 0:
            return False
    return True

primeL = [n for n in randL if is_prime(n)]
print(primeL)
print("Count of prime numbers:", len(primeL))
```

```python
#Q3(v)
from collections import Counter
freq = Counter(randL)
most_common_num, frequency = freq.most_common(1)[0]
print("Most frequent number:", most_common_num)
print("Number of times it occurs:", frequency)
```

## Q4 - Set operations

```python
#Q4 - creating sets A and B
digits = [int(d) for d in roll_no[:8]]   # the 8 original digits (before multiplying by 10)
A = set(d*7 for d in digits)
B = set(d*9 for d in digits)
print("A =", A)
print("B =", B)
```

```python
#Q4(vi)
print(A.union(B))
```

```python
#Q4(vii)
print(A.intersection(B))
```

```python
#Q4(viii)
print(A.difference(B))
print(B.difference(A))
# difference() only shows values that are in one set but not the other, in the given order (A-B or B-A)
# symmetric_difference() combines both of these results together into a single set
```

```python
#Q4(ix)
print(A.symmetric_difference(B))
```

```python
#Q4(x)
print(A.issubset(B))
print(B.issuperset(A))
```

```python
#Q4(xi)
X = int(input("Enter a value X: "))
A.discard(X)   # discard() is safer than remove() because it won't throw an error if X is not in the set
print(A)
```

## Q5 - Dictionary operations

```python
#Q5 - creating my dictionary
my_dict = {
    "name": "Himantveer Kaur",
    "roll_no": "1024170271",
    "branch": "COPC",
    "age": 20,
    "city": "Bathinda"
}
print(my_dict)
```

```python
#Q5(i)
my_dict["location"] = my_dict.pop("city")   # pop() removes "city" and returns its value, which we store under "location"
print(my_dict)
```

```python
#Q5(ii)
my_dict["cgpa"] = 9.89
print(my_dict)
```

```python
#Q5(iii)
my_dict["age"] += 1
print(my_dict)
```

```python
#Q5(iv)
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

dict_pop.pop("branch")
del dict_del["branch"]

print(dict_pop)
print(dict_del)
# pop() removes the key and also returns its value, while del just removes the key and returns nothing
```

```python
#Q5(v)
for key, value in my_dict.items():
    print(key, "->", value)
```

```python
#Q5(vi)
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")
```

```python
#Q5(vii)
friend_dict = {
    "name": "Ansh Bindal",
    "roll_no": "1024170305",
    "branch": "CSE",
    "age": 21,
    "city": "Ludhiana"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
# when both dictionaries share a key, the value from the second dictionary (friend_dict) overwrites the first
```

```python
#Q5(viii)
string_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print(string_dict)
```
