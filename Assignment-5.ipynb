### Q2

```python
# Q2 a)
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original:", arr)
print("Reversed:", arr[::-1])
```

```python
# Q2 b) i.
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = counts.max()

for v, c in zip(values, counts):
    if c == max_count:
        print("Most frequent value:", v)
        print("Count:", c)
        print("Indices:", np.where(x == v)[0])
```

```python
# Q2 b) ii.
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = counts.max()

for v, c in zip(values, counts):
    if c == max_count:
        print("Most frequent value:", v)
        print("Count:", c)
        print("Indices:", np.where(y == v)[0])
```

### Q4

```python
# Q4
Himantveer_Kaur = np.linspace(10, 100, 25)
print(Himantveer_Kaur)

print("Dimensions:", Himantveer_Kaur.ndim)
print("Shape:", Himantveer_Kaur.shape)
print("Total elements:", Himantveer_Kaur.size)
print("Data type:", Himantveer_Kaur.dtype)
print("Total bytes:", Himantveer_Kaur.nbytes)
```

```python
# Q4 transpose using reshape
col = Himantveer_Kaur.reshape(25, 1)
print("Shape after reshape:", col.shape)
print(col)
```

```python
# Q4 using T attribute
print("Shape using T:", Himantveer_Kaur.T.shape)
print("Same as original?", np.array_equal(Himantveer_Kaur.T, Himantveer_Kaur))
```

No, we cannot get the same result with `T` here. A 1-D array has only one axis, so `T` returns the same array with shape `(25,)`. Reshape is needed to make it a 2-D array of shape `(25, 1)` (or `(1, 25)`); after that `T` would work.
